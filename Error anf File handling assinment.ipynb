### Q4 
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
# Q4 solution code here


with open("numbers.txt", "w") as file:
    for number in range(1, 6):
        file.write(str(number) + "\n")

print("Numbers written successfully.")
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here


with open("numbers.txt", "r") as file:
    numbers = file.readlines()

total = 0

for number in numbers:
    total += int(number.strip())

print("Sum:", total)
```

### Q13 (Easy)
Aisi file open karne ki koshish karein jo exist nahi karti (`non_existent_file.txt`). `FileNotFoundError` ko `try-except` se handle karein.

```python
# Q13 solution code here


try:
    with open("non_existent_file.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Error: File does not exist.")
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here


def validate_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    return "Age is valid."


try:
    age = int(input("Enter your age: "))
    print(validate_age(age))
except ValueError as e:
    print("Error:", e)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here


file = None

try:
    file = open("data.txt", "r")
    content = file.read()
    print(content)

except FileNotFoundError:
    print("Error: File nahi mili.")

finally:
    if file is not None:
        file.close()
        print("File closed successfully.")
```
